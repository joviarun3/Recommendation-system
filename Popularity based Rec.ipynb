```python
#Importing libraries
import pandas as pd
import numpy as np
```

```python
#Read the movie content
i_cols = ['movie id', 'movie title' ,'release date','video release date', 'IMDb URL', 'unknown', 'Action', 'Adventure',
    'Animation', 'Children\'s', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy',
    'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']
items = pd.read_csv('ml-100k/u.item', sep='|', names=i_cols, encoding='latin-1')
```

```python
items
```

```python
movie_content = items[['movie id','movie title','Action', 'Adventure',
    'Animation', 'Children\'s', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy',
    'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']]
movie_content
```

```python
#create rating table from data
r_cols = ['user_id', 'movie_id', 'rating', 'unix_timestamp']
ratings = pd.read_csv('ml-100k/u.data', sep='\t', names=r_cols, encoding='latin-1')
```

```python
ratings
```

```python
# List of genre columns
genres = ['Action', 'Adventure', 'Animation', "Children's", 'Comedy', 'Crime','Documentary', 'Drama', 'Fantasy', 'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']
```

```python
# Let's cerate empty Dictionary to hold DataFrames for each genre
genre_dfs = {}
```

```python
for i in genres:
    genre_dfs[i] = movie_content[movie_content[i] == 1][['movie id', 'movie title']]
```

```python
len(genre_dfs)
```

```python
#checking
genre_dfs[genres[0]]
```

```python
#Cross checking to access the tuples.
genre_dfs.items()
```

```python
#create empty dictionary for to store df for top5 movies in each genre
top5_by_genre = {}

for genre, df in genre_dfs.items():     
# genre is key & df is value in dictionary
# .items() returns all pairs from the dictionary as tuples:

    # first we will bring ratings for movies by merging/vlook up in excel
    merged = pd.merge(df, ratings, left_on='movie id', right_on='movie_id')
    
    # Group ratings by movie id and title. Each movie has different ratings by diff users in groups now.
    grouped = merged.groupby(['movie id', 'movie title'])['rating']

    #Find mean rating for each movie grouped. we can't just go with mzx rating if it is an outlier (e.g) only one raing in top.
    avg_rating_per_movie = grouped.mean()

    #Convert groupings back into a DataFrame
    avg_ratings = pd.DataFrame(avg_rating_per_movie)

    #Sort movies by average rating (in decending order)
    avg_ratings_sorted = avg_ratings.sort_values(by='rating', ascending=False)

    # Select top 5 movies for this genre (it loops through for loop for each genre & df)
    top5 = avg_ratings_sorted.head(5)
    
    #storing in dictionary..
    top5_by_genre[genre] = top5
```
